### Import Pandas

```python
import pandas as pd
```

### User Input: Beam Geometry and Applied Loads

```python

beam_length = float(input("Enter beam length (ft): "))
support_A = float(input("Enter location of Support A (ft): "))
support_B = float(input("Enter location of Support B (ft): "))

num_loads = int(input("Enter number of loads: "))

load_x = []
load_y = []
locations = []

for i in range(num_loads):
    print("\nLoad", i+1)
    
    Fx = float(input("Enter horizontal load (lb) (use 0 if none): "))
    Fy = float(input("Enter vertical load (lb, downward positive): "))
    location = float(input("Enter location of load (ft): "))
    
    load_x.append(Fx)
    load_y.append(Fy)
    locations.append(location)

loads = pd.DataFrame({"Fx (lb)": load_x, "Fy (lb)": load_y, "Location (ft)": locations})

print("\nLoads entered:")
print(loads)
```

### Calculate Support Reactions Using Static Equilibrium

```python
total_Fy = loads["Fy (lb)"].sum()

total_Fx = loads["Fx (lb)"].sum()

moment_about_A = (loads["Fy (lb)"] * (loads["Location (ft)"] - support_A)).sum()

reaction_By = moment_about_A / (support_B - support_A)
reaction_Ay = total_Fy - reaction_By

reaction_Ax = -total_Fx

print("\nResults:")
print("Reaction Ax =", round(reaction_Ax, 2), "lb")
print("Reaction Ay =", round(reaction_Ay, 2), "lb")
print("Reaction By =", round(reaction_By, 2), "lb")
```

### Verify Equilibrium Conditions (ΣFx, ΣFy, ΣM)

```python
force_x_check = reaction_Ax + total_Fx
force_y_check = reaction_Ay + reaction_By - total_Fy
moment_check = reaction_By * (support_B - support_A) - moment_about_A

print("\nCheck Results:")
print("Sum of Forces in X =", round(force_x_check, 4))
print("Sum of Forces in Y =", round(force_y_check, 4))
print("Sum of Moments =", round(moment_check, 4))

tolerance = 0.01

if abs(force_x_check) < tolerance and abs(force_y_check) < tolerance and abs(moment_check) < tolerance:
    print("The beam IS in equilibrium")
else:
    print("The beam is NOT in equilibrium")
```

### Example Problem

```python
beam_length = 20
support_A = 0
support_B = 20

loads = pd.DataFrame({"Fx (lb)": [0, 0], "Fy (lb)": [500, 800], "Location (ft)": [6, 14]})

total_Fy = loads["Fy (lb)"].sum()
total_Fx = loads["Fx (lb)"].sum()

moment_about_A = (loads["Fy (lb)"] * (loads["Location (ft)"])).sum()

reaction_By = moment_about_A / (support_B - support_A)
reaction_Ay = total_Fy - reaction_By
reaction_Ax = -total_Fx

force_x_check = reaction_Ax + total_Fx
force_y_check = reaction_Ay + reaction_By - total_Fy
moment_check = reaction_By * (support_B - support_A) - moment_about_A

print("\nExample Problem")
print("Reaction Ax =", round(reaction_Ax, 2), "lb")
print("Reaction Ay =", round(reaction_Ay, 2), "lb")
print("Reaction By =", round(reaction_By, 2), "lb")

if abs(force_x_check) < 0.01 and abs(force_y_check) < 0.01 and abs(moment_check) < 0.01:
    print("The beam IS in equilibrium")
else:
    print("The beam is NOT in equilibrium")
```
